```python
import os
from dotenv import load_dotenv
load_dotenv() #initialize dotenv

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
```

```python
from langchain.agents import create_agent

def get_weather(city: str) -> str:
    """In a real-world scenario, you would call a weather API to get the actual weather data."""
    return f"The current weather in {city} is sunny with a temperature of 25°C."

agent=create_agent(
    model="gpt-5",
    tools=[get_weather],
    system_prompt="Hey, you are helful assistant",
)
agent.invoke({
    "messages": [{"role": "user", "content": "What is the weather in Mumbai?"}]
})
```
